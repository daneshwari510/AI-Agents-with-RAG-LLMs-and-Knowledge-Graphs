## Chunking and Tokenization

## Without Overlap

```python
from IPython.display import HTML
from langchain_text_splitters import TokenTextSplitter


def return_splits(text, splitter):
    docs = splitter.create_documents([text])
    chunks = [doc.page_content for doc in docs]
    for i, chunk in enumerate(chunks):
        print(f"Chunk {i+1}:\n{chunk}\n")


text = '''To be or not to be, that is the question.
Whether tis nobler in the mind to suffer
The slings and arrows of outrageous fortune
Or to take arms against a sea of troubles,
And by opposing, end them. To die, to sleep
No more, and by a sleep to say we end,
The heartache and the thousand natural shocks
That flesh is heir to, tis a consummation
Devoutly to be wished.'''

return_splits(text, TokenTextSplitter(chunk_size=50, chunk_overlap=0))
```

## Easy Overlap

```python
return_splits(text, TokenTextSplitter(chunk_size=50, chunk_overlap=5))


```

## Character aware splitter (fails overlap)



```python
from  langchain_text_splitters import CharacterTextSplitter
token_splitter_overlap = return_splits(text, CharacterTextSplitter(separator="\n", chunk_size=50, chunk_overlap=15))
```

```python
from  langchain_text_splitters import CharacterTextSplitter
token_splitter_overlap = return_splits(text, CharacterTextSplitter(separator=" ", chunk_size=50, chunk_overlap=15))
```

## Recursive character splitting

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = '''To be or not to be, that is the question. Whether tis nobler in the mind to suffer. The slings and arrows of outrageous fortune
Or to take arms against a sea of troubles,
And by opposing, end them. To die, to sleep
No more, and by a sleep to say we end,
The heartache and the thousand natural shocks
That flesh is heir to, tis a consummation
Devoutly to be wished.'''

return_splits(text,RecursiveCharacterTextSplitter(
        separators = ["\n\n", "\n", " ", ""],
        chunk_size=50,
        chunk_overlap=10
    ))
```
